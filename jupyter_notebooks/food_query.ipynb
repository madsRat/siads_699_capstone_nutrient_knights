```python
import pandas as pd

def search_food(food_name, food_df):
    """Search for an exact food item and prioritize high-quality data sources."""
    food_name = food_name.upper()  # Convert input to uppercase to match USDA format
    food_df['description_upper'] = food_df['description'].str.upper()  # Create a case-insensitive column

    # Exact match filtering
    exact_matches = food_df[
        (food_df['description_upper'] == food_name) # |  # Exact match
        # (food_df['description_upper'].str.startswith(food_name + ",")) |  # Handles "APPLE, raw"
        # (food_df['description_upper'].str.endswith("," + food_name))  # Handles "RAW, APPLE"
    ]
    
    return exact_matches[['fdc_id', 'description', 'data_type']]

# Load CSV file
food_df = pd.read_csv("food.csv") # change to your file path

query = 'peanut'

results = search_food(query, food_df)

if results.empty:
    print(f"No results found for '{query}'.")
else:
    print("Refined Matching Foods:")
    print(results)
```
